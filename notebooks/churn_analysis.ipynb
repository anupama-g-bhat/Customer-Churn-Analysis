```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the  dataset
df = pd.read_csv('data\WA_Fn-UseC_-Telco-Customer-Churn.csv')
```

```python
# Checking the columns
df.columns
```

```python
df.head()
```

```python
df.shape
```

```python
# Finding missing values
df.isnull().sum()
```

```python
# Data type information
df.info()
```

```python
df['Churn'].value_counts()
```

```python
plt.figure(figsize=(6, 6))
plt.pie(df['Churn'].value_counts(), 
        labels=['Retained', 'Churned'],
        colors=['green', 'red'],
        autopct='%1.1f%%',  # shows percentage on pie
        explode=(0, 0.1))  # slightly explode churned slice
plt.title('Overall Customer Churn Distribution')
plt.savefig('overall_churn.png', dpi=150)
plt.show()
```

```python
# Convert TotalCharges from text to number
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors ='coerce')

# Verify
df['TotalCharges'].dtype
```

```python
# Convert churn from yes/no to 1/0

df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

print(f"Churn dtype: {df['Churn'].dtype}")
print(f"Overall Churn Rate: {df['Churn'].mean()*100:.2f}%")
```

```python
# Churn rate by Contract type
contract_churn = df.groupby('Contract')['Churn'].mean()* 100
print(contract_churn)
```

```python
plt.figure(figsize=(8, 5))

contract_churn.plot(kind='bar', 
                    color=['red','orange','green'],
                    edgecolor='black')

plt.title('Churn Rate by Contract Type (%)')
plt.xlabel('Contract Type')
plt.ylabel('Churn Rate (%)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('contract_churn.png', dpi=150)
plt.show()
```

```python
# Average tenure of churned vs not churned
tenure_churn = df.groupby('Churn')['tenure'].mean()
print(tenure_churn)
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='Churn', y='tenure', data=df,
            palette=['green','red'])
plt.title('Tenure Distribution: Churned vs Retained')
plt.xlabel('Churn (0=Stayed, 1=Left)')
plt.ylabel('Tenure (Months)')
plt.savefig('tenure_churn.png', dpi=150)
plt.show()
```

```python
charges_churn = df.groupby('Churn')['MonthlyCharges'].mean()
print(charges_churn)
```

```python
plt.figure(figsize=(8, 5))
sns.violinplot(x='Churn', y='MonthlyCharges', 
               data=df, palette=['green','red'])
plt.title('Monthly Charges Distribution: Churned vs Retained')
plt.xlabel('Churn (0=Stayed, 1=Left)')
plt.ylabel('Monthly Charges ($)')
plt.savefig('charges_churn.png', dpi=150)
plt.show()
```

```python
internet_churn = df.groupby('InternetService')['Churn'].mean() * 100
print(internet_churn)
```

```python
plt.figure(figsize=(8, 5))

internet_churn.plot(kind='bar', 
                    color=['green','red','orange'],
                    edgecolor='black')

plt.title('Churn Rate by Internet Service (%)')
plt.xlabel('InternetService')
plt.ylabel('Churn Rate (%)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig('internet_churn.png', dpi=150)
plt.show()
```

```python
Senior_citizen_churn = df.groupby('SeniorCitizen')['Churn'].mean() * 100
print(Senior_citizen_churn)
```

```python
plt.figure(figsize=(8, 5))

Senior_citizen_churn.plot(kind='bar', 
                    color=['green','red'],
                    edgecolor='black')

plt.title('Churn Rate by Senior Citizen (%)')
plt.xticks([0, 1], ['Non-Senior', 'Senior Citizen'], rotation=0)  
plt.xlabel('Customer Type')       
plt.ylabel('Churn Rate (%)')       
plt.tight_layout()
plt.savefig('Senior_citizen_churn.png', dpi=150)
plt.show()
```

```python
# Customer Lifetime Value approximation
df['CLV'] = df['tenure'] * df['MonthlyCharges']

clv_churn = df.groupby('Churn')['CLV'].mean()
print("Average Customer Lifetime Value:")
print(f"Retained customers: ${clv_churn[0]:,.2f}")
print(f"Churned customers : ${clv_churn[1]:,.2f}")
```

```python
# Visualize CLV
plt.figure(figsize=(8,5))
clv_churn.plot(kind='bar',
               color=['green','red'],
               edgecolor='black')
plt.title('Average Customer Lifetime Value: Retained vs Churned')
plt.xticks([0,1], ['Retained','Churned'], rotation=0)
plt.xlabel('Customer Type')
plt.ylabel('Average CLV ($)')
plt.tight_layout()
plt.savefig('clv_analysis.png', dpi=150)
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
fig.suptitle('Customer Churn Analysis Dashboard\nFuture Interns — Data Science & Analytics', 
             fontsize=16, fontweight='bold')

# Chart 1 — Contract vs Churn
contract_churn.plot(kind='bar', 
                    color=['red','orange','green'],
                    edgecolor='black',
                    ax=axes[0,0])
axes[0,0].set_title('Churn Rate by Contract Type (%)')
axes[0,0].set_xlabel('Contract Type')
axes[0,0].set_ylabel('Churn Rate (%)')
axes[0,0].tick_params(axis='x', rotation=0)

# Chart 2 — Tenure Boxplot
sns.boxplot(x='Churn', y='tenure', data=df,
            palette=['green','red'], ax=axes[0,1])
axes[0,1].set_title('Tenure Distribution: Churned vs Retained')
axes[0,1].set_xlabel('Churn (0=Stayed, 1=Left)')
axes[0,1].set_ylabel('Tenure (Months)')

# Chart 3 — Monthly Charges Violin
sns.violinplot(x='Churn', y='MonthlyCharges', 
               data=df, palette=['green','red'],
               ax=axes[0,2])
axes[0,2].set_title('Monthly Charges: Churned vs Retained')
axes[0,2].set_xlabel('Churn (0=Stayed, 1=Left)')
axes[0,2].set_ylabel('Monthly Charges ($)')

# Chart 4 — Internet Service vs Churn
internet_churn.plot(kind='bar', 
                    color=['green','red','orange'],
                    edgecolor='black',
                    ax=axes[1,0])
axes[1,0].set_title('Churn Rate by Internet Service (%)')
axes[1,0].set_xlabel('Internet Service')
axes[1,0].set_ylabel('Churn Rate (%)')
axes[1,0].tick_params(axis='x', rotation=0)

# Chart 5 — Senior Citizen vs Churn
Senior_citizen_churn.plot(kind='bar', 
                    color=['green','red'],
                    edgecolor='black',
                    ax=axes[1,1])
axes[1,1].set_title('Churn Rate by Senior Citizen (%)')
axes[1,1].set_xticks([0, 1])
axes[1,1].set_xticklabels(['Non-Senior', 'Senior Citizen'])
axes[1,1].set_xlabel('Customer Type')
axes[1,1].set_ylabel('Churn Rate (%)')

# Chart 6 — Overall Churn Pie
axes[1,2].pie(df['Churn'].value_counts(), 
              labels=['Retained', 'Churned'],
              colors=['green', 'red'],
              autopct='%1.1f%%',
              explode=(0, 0.1))
axes[1,2].set_title('Overall Customer Churn Distribution')

plt.tight_layout()
plt.savefig('churn_dashboard.png', dpi=150)
plt.show()
```
